```python
# we are goin to remove the unnecessary spaces
book = [x.strip() for x in book.x] # remove both leading and the trailing spaces
book = [x for x in book if x] # removes empty strings, because they are considered in Python as False
book[0:10]
```

```python
#Joining the list into one string/text and
text = ' '.join(book)
text
```

```python
#Punctuation
no_punc_text = text.translate(str.maketrans('','', string.punctuation))
#with arguments (x.y,z) where 'x' and'y'
#must be equal-length strings and charcaters in 'x' are replaced by characters in'y'.
# 'z' is a  (string.punctuation here)
no_punc_text
```

```python
# Alternate code for removing Punctuation
#no_punc_text = [word for word in no_punc_text if word.isalnum()]
#no_punc_text
```

```python
import nltk
nltk.download('punkt')
```

```python
#Tokenization
from nltk.tokenize import word_tokenize
text_tokens = word_tokenize(no_punc_text)
print(text_tokens[0:50])
```

```python
len(text_tokens)
```

```python
#Normalize the data
lower_words = [x.lower() for x in text_tokens]
print(lower_words[0:25])
```

```python
#remove Stopwords
from nltk.corpus import stopwords
nltk.download('stopwords')
```

```python
my_stop_words = stopwords.words('english')
my_stop_words
```

```python
my_stop_words.append('the') # if u do want to delete ceratin stop words u can append them
no_stop_tokens = [word for word in lower_words if not word in my_stop_words]
print(no_stop_tokens[0:40])
```

```python
len(no_stop_tokens)
```

```python
len(lower_words)
```

```python
#Stemming
from nltk.stem import PorterStemmer
ps = PorterStemmer()
```

```python
stemmed_tokens = [ps.stem(word) for word in no_stop_tokens]
print(stemmed_tokens[0:40])
```

```python
for w in no_stop_tokens:
  print(w, ':', ps.stem(w))
```

```python
## import the libraries
from spacy.lemmatizer import Lemmatizer
from spacy.lookups import Lookups

##lemmatization
doc = nlp(u'I love coding and writing')
for word in doc:
  print(word.text, ">", word.lemma_)
```

```python
#define a function to plot word cloud
def plot_cloud(wordcloud):
  #set figure size
  plt.figure(figsize=(40, 30))
  #display image
  plt.imshow(wordcloud)
  #No axis details
  plt.axis("off");
```
